```python
df2 = df.groupby('Continent').mean(numeric_only=True).sort_values(by="2022 Population", ascending=False)
df2
```

```python
df3 = df2.transpose()
df3
```

```python
# Select population columns and reverse order so years go chronologically (1970 -> 2022)
pop_cols = ['1970 Population', '1980 Population', '1990 Population', 
            '2000 Population', '2010 Population', '2015 Population', 
            '2020 Population', '2022 Population']

# Transpose df2 for plotting years on the x-axis
df2_trend = df2[pop_cols].T

plt.figure(figsize=(12, 6))
sns.lineplot(data=df2_trend, markers=True, dashes=False)
plt.title('Average Population Trend by Continent (1970–2022)', fontsize=14)
plt.xlabel('Year', fontsize=12)
plt.ylabel('Average Population', fontsize=12)
plt.xticks(rotation=45)
plt.legend(title='Continent')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
